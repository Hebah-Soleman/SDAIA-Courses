# Custom Object Detection: Annotate with Roboflow, Train with YOLO

**Goal of this exercise:** learn the full workflow of building a custom object
detection dataset from scratch and training a model on it.

You will:
1. Download the [Construction Site Safety](https://drive.google.com/file/d/1RSr3Q3bG6MnoMqtAynS7_1lnvcAkkn-a/view?usp=sharing) 
2. Upload and **manually annotate** ~20–30 images in Roboflow (draw bounding boxes)
3. Export the annotated dataset in YOLO format
4. Train a YOLOv8 model on it with `ultralytics`
5. Evaluate the model and run inference on new images



## Step 1 — Set Up the Environment

Install the two libraries we need:

## Step 2 — Create a Roboflow Account and Project

Do this in your browser (not code):

1. Go to [https://roboflow.com](https://roboflow.com) and sign up for a free account.
2. Click **Create New Project**.
   - **Project Type:** Object Detection
   - **Annotation Group:** name it after what you're detecting, e.g. `objects`
3. Note your:
   - **Workspace ID** (visible in the project URL)
   - **Project name** (the slug in the URL)
4. Go to your [Roboflow Settings → API Key](https://app.roboflow.com/settings/api) page and copy your **Private API Key**.

Paste your credentials below (never commit real API keys to a public repo —

## Step 3 — Export the Annotated Dataset in YOLO Format

Once your version is generated, download it in **YOLOv8** format directly
into this notebook's environment.


## Step 4 — Train a YOLOv8 Model

We start from a pretrained **nano** checkpoint (`yolov8n.pt`) — smallest and
fastest YOLOv8 variant, ideal for a small dataset and short training runs.

For a dataset this size, ~50–100 epochs is usually plenty. Adjust
`epochs`/`imgsz`/`batch` if training is too slow on your hardware.


## Step 5 — Evaluate the Model

Run validation on the held-out `valid` split to see precision, recall, and
mAP (mean Average Precision).


> With only ~20–30 training images, don't expect a very high mAP — that's
> expected and fine. The goal here is to see the *pipeline* work, not to
> hit production-level accuracy. If you want better numbers: add more
> annotated images, use a larger YOLOv8 variant (`yolov8s.pt`), or increase
> augmentation.


## Step 6 — Run Inference on New Images

Try the trained model on a few images it hasn't seen (e.g. from the `test`
split, or brand-new photos you take now).


## Step 7 — Save and Export the Trained Model

The best weights are saved automatically during training. You can also
export to other formats (ONNX, TFLite, etc.) for deployment.
